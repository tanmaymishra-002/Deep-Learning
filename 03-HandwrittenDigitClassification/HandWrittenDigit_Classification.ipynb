```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense,Flatten
```

```python
(X_train,y_train),(X_test,y_test) = keras.datasets.mnist.load_data()
```

you can observe that the values are in range of 0 to 255 but we want the values between 0 to 1 why?-> because the more similar the values are the faster the weights could be calculated so
we will divide all the data with the highest value i.e 255 as a result the whole data will be in range of 0 to 1

```python
X_train=X_train/255
```

```python
X_test=X_test/255
```

So now we want to build the neural network right ?

but, there is a problem why ?

our data is in 28x28 form
but we want it in 1x784 form to give it in input features to our neural network right ?

so for that we will use Flatten

```python
model = Sequential()
model.add(Flatten(input_shape=(28,28)))
model.add(Dense(128,activation='relu')) #We did not add number of input features because the output of flatten will eventually be added there
model.add(Dense(10,activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(loss='sparse_categorical_crossentropy',optimizer='Adam')
```

```python
model.fit(X_train,y_train,epochs=10,validation_split=0.2)
```

```python
y_prob=model.predict(X_test)
```

```python
y_pred=y_prob.argmax(axis=1)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```
